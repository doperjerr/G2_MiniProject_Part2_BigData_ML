## 1. ปัญหาและคำถามที่สนใจ
**ปัญหา/คำถามที่สนใจคืออะไร**

รูปแบบการเคลื่อนที่ทำทาง (Movement Archetypes) แตกต่างกันอย่างไร และทีมที่ชนะมีสไตล์การเล่นอย่างไร

**ปัญหา/คำถามนี้มีความสำคัญอย่างไร และใครจะนำผลไปใช้**

- ความสำคัญ:
ช่วยเปลี่ยนข้อมูลพิกัดการเคลื่อนที่ดิบปริมาณมหาศาล (Big Data) ให้กลายเป็นยุทธวิธีเชิงพื้นที่ (Tactical Insights) โดยทำให้เห็นโปรไฟล์รูปแบบการบุกและการเคลื่อนที่ทำทางสากลของทีมในนัดที่ประสบความสำเร็จ (Winning Matches)เข้าใจโครงสร้างเกมบุกของทีมที่ประสบความสำเร็จเพื่อวางแนวทางตั้งรับดักทางได้อย่างตรงจุด

- ผู้นำผลไปใช้:  
  -โค้ชและทีมงานสเกาต์ (Coaches & Tactical Analysts): นำโปรไฟล์สไตล์การเล่นและรูปแบบการเคลื่อนที่ทำทาง (Movement Archetypes) ของทีมที่ชนะ ไปศึกษาวิเคราะห์เพื่อวางแผนการซ้อม วางแนวทางตั้งรับดักทาง และจัดแท็กติกรับมือคู่แข่งก่อนลงแข่งขันจริง

  -นักกีฬา (Players): นำภาพจำลองเส้นทางการเคลื่อนที่ (เช่น 2D Pitch Map) ไปศึกษาและทำความเข้าใจ เพื่ออ่านเกมล่วงหน้า อ่านจังหวะการยืนปิดช่องทางส่งบอล หรือการพยายามเลี้ยงบอลทำทางในสนาม

   -นักวิเคราะห์ข้อมูลสโมสร (Sports Data Scientists): นำ Data Pipeline และโมเดลการจัดกลุ่มสไตล์การเคลื่อนที่นี้ไปใช้ประเมินประสิทธิภาพยุทธวิธีเชิงพื้นที่ และเปรียบเทียบสไตล์การเล่นของนักเตะรวมถึงทีมสโมสรได้อย่างมีมาตรฐาน

## 2. Data Description
* แหล่งข้อมูล (ลิงก์) และช่วงเวลาของข้อมูล

  แหล่งข้อมูล: [StatsBomb Open Data Repository (GitHub Raw JSON)](https://github.com/hudl/open-data)

  ช่วงเวลาของข้อมูล: ครอบคลุมการแข่งขันฟุตบอลทัวร์นาเมนต์หลักระดับโลกตั้งแต่ปี 2004 – 2024 (เช่น FIFA World Cup, UEFA Euro, UEFA Champions League)

---

* จำนวนแถว จำนวนคอลัมน์ และความหมายของแต่ละคอลัมน์ที่ใช้ (data dictionary)

  ประมวลผลโดยนำตารางดิบ 2 ตาราง มาเชื่อมโยงกันด้วยคอลัมน์ match_id (Inner Join):
1. ตาราง matches  ใช้ดึงผลการแข่งขันเพื่อคำนวณอัตราการชนะ (Win Rate)
2. ตาราง events ใช้ดึงพิกัดการเคลื่อนที่และประเภทเหตุการณ์

* จำนวนแถว (Rows): ประมาณ 9M แถว (เหตุการณ์ทั้งหมดจากทุกแมตช์)
* จำนวนคอลัมน์ที่ใช้: 8 คอลัมน์หลัก

1. matches.json -> match_id (Integer): รหัสอ้างอิงการแข่งขัน (ใช้เป็น Key ในการ Merge)
2. matches.json -> home_score, away_score (Integer): ผลการแข่งขันเพื่อคำนวณอัตราการชนะ (Win Rate) รายทีม
3. events.json -> match_id (Integer): รหัสอ้างอิงการแข่งขัน (ใช้เป็น Key ในการ Merge)
4. events.json -> team.name (String): ชื่อทีมที่ทำกิจกรรม (ใช้กรองกลุ่ม High Win-Rate Teams)
5. events.json -> type.name (String): ประเภทเหตุการณ์ คัดเฉพาะเหตุการณ์เชิงพื้นที่ (Pass, Carry)
6. events.json -> location (Array [x, y]): พิกัดจุดเริ่มต้นบนสนาม 120 x 80 หลา (start_x, start_y)
7. events.json -> pass.end_location / carry.end_location (Array [x, y]): พิกัดจุดสิ้นสุดบนสนาม 120 x 80 หลา (end_x, end_y)

---

* ข้อมูล 1 แถว แทนอะไร (unit of analysis)

  ข้อมูลดิบ 1 แถว แทน "1 เหตุการณ์การเคลื่อนที่เชิงพื้นที่ (Single Spatial Movement Event)" ของผู้เล่นในสนามในแมตช์นั้นๆ (เช่น การส่งบอล 1 ครั้ง หรือการพาบอล 1 ครั้ง) ที่บันทึกพิกัดจุดเริ่มต้นและจุดสิ้นสุดไว้อย่างชัดเจน

## 3. Data Loading

**------------------------------------ห้ามรัน----------------------------------------**

```python
# Load data
# Merge data if necessary (ระบุ key ที่ใช้ และตรวจจำนวนแถวก่อนและหลัง merge)
```

```python
import pandas as pd
import numpy as np
import requests
from tqdm import tqdm
import concurrent.futures
import time # เพิ่ม time

GITHUB_BASE_URL = "https://raw.githubusercontent.com/statsbomb/open-data/master/data"

# =================================================================
# 1. ดึงรายชื่อ Competitions และ Matches ทั้งหมดที่มีบน GitHub
# =================================================================
print("1. กำลังดึงรายชื่อการแข่งขันและแมตช์ทั้งหมดจาก GitHub...")
comp_url = f"{GITHUB_BASE_URL}/competitions.json"
competitions = requests.get(comp_url).json()

all_match_ids = []
matches_list = [] # เพิ่ม list สำหรับเก็บข้อมูลตาราง Matches
for comp in competitions:
    c_id, s_id = comp['competition_id'], comp['season_id']
    matches_url = f"{GITHUB_BASE_URL}/matches/{c_id}/{s_id}.json"
    res = requests.get(matches_url)
    if res.status_code == 200:
        for m in res.json():
            all_match_ids.append(m['match_id'])

            # เก็บข้อมูลลง matches_list ทันที
            home_team = m["home_team"]["home_team_name"]
            away_team = m["away_team"]["away_team_name"]
            home_score = m["home_score"]
            away_score = m["away_score"]

            winner = None
            if home_score > away_score:
                winner = home_team
            elif away_score > home_score:
                winner = away_team

            matches_list.append({
                "match_id": m["match_id"],
                "home_team": home_team,
                "away_team": away_team,
                "winner": winner
            })

all_match_ids = list(set(all_match_ids))
df_matches = pd.DataFrame(matches_list).drop_duplicates(subset=["match_id"]) # สร้างตาราง df_matches
print(f"พบแมตช์ทั้งหมดบน GitHub: {len(all_match_ids):,} นัด")

# =================================================================
# 2. ฟังก์ชันดึงและแปลงข้อมูล Events ทุกประเภทที่มีพิกัด X, Y
# =================================================================
def fetch_all_events(match_id, retries=3): # เพิ่ม retries
    url = f"{GITHUB_BASE_URL}/events/{match_id}.json"

    # เพิ่มลูปสำหรับ retry
    for attempt in range(retries):
        try:
            res = requests.get(url, timeout=10) # เพิ่ม timeout ป้องกันค้าง
            if res.status_code != 200:
                return None
            events = res.json()
            break # ถ้าโหลดสำเร็จ ให้ออกจากลูป
        except Exception as e:
            if attempt < retries - 1:
                time.sleep(2) # รอ 2 วินาทีแล้วลองใหม่
                continue
            else:
                return None # ถ้าลองครบแล้วยังไม่ได้ ให้คืนค่า None

    extracted = []
    target_types = ['Pass', 'Carry'] # คัดเฉพาะ Pass กับ Carry เลย ตามที่คุณต้องการ
    for ev in events:
        event_type = ev.get('type', {}).get('name')
        if event_type in target_types:
            team_name = ev.get('team', {}).get('name')
            player_name = ev.get('player', {}).get('name')
            loc = ev.get('location')
            if not loc or len(loc) < 2:
                continue
            start_x, start_y = loc[0], loc[1]
            end_x, end_y = start_x, start_y

            if event_type == 'Pass' and 'pass' in ev:
                end_loc = ev['pass'].get('end_location')
                if end_loc and len(end_loc) >= 2:
                    end_x, end_y = end_loc[0], end_loc[1]
            elif event_type == 'Carry' and 'carry' in ev:
                end_loc = ev['carry'].get('end_location')
                if end_loc and len(end_loc) >= 2:
                    end_x, end_y = end_loc[0], end_loc[1]

            extracted.append({
                'match_id': match_id,
                'team': team_name,
                'player': player_name,
                'type': event_type,
                'start_x': start_x,
                'start_y': start_y,
                'end_x': end_x,
                'end_y': end_y
            })
    return extracted

# =================================================================
# 3. ยิงดึงทุกแมตช์ด้วย Parallel Processing (ลดเหลือ 10 Threads)
# =================================================================
print(f"2. กำลังดาวน์โหลดข้อมูลจากทุกแมตช์ ({len(all_match_ids):,} นัด) เข้า RAM...")
all_data = []

# ลด max_workers ลงเพื่อป้องกัน Connection Reset
with concurrent.futures.ThreadPoolExecutor(max_workers=10) as executor:
    results = list(tqdm(executor.map(fetch_all_events, all_match_ids), total=len(all_match_ids)))

for res in results:
    if res:
        all_data.extend(res)

df_all_movements = pd.DataFrame(all_data)
print(f"\nดึงข้อมูลทุกแมตช์สำเร็จ! ได้จำนวนแถวรวมทั้งหมดใน RAM: {len(df_all_movements):,} แถว")

# =================================================================
# 4. รวม 2 ตารางเข้าด้วยกัน (Merge)
# =================================================================

# แมปชื่อทีม
ev_pairs = df_all_movements[["match_id", "team"]].drop_duplicates()
ev_pairs["team"] = ev_pairs["team"].astype(str)

pairs = ev_pairs.merge(df_matches[["match_id", "home_team", "away_team", "winner"]],
                       on="match_id", how="left", validate="m:1").reset_index(drop=True)

pairs["canon"] = np.where((pairs.team == pairs.home_team) | (pairs.team == pairs.away_team), pairs.team, np.nan)

hints = {}
for _round in range(10):
    changed = False
    todo = pairs.loc[pairs.canon.isna(), "match_id"].unique()
    for mid, gg in pairs[pairs.match_id.isin(todo)].groupby("match_id"):
        home, away = gg.home_team.iloc[0], gg.away_team.iloc[0]
        for i in gg.index[gg.canon.isna()]:
            cands = hints.get(pairs.at[i, "team"], set()) & {home, away}
            if len(cands) == 1:
                pairs.at[i, "canon"] = next(iter(cands))
                changed = True

        gg = pairs.loc[gg.index]
        unres, res = gg[gg.canon.isna()], gg[gg.canon.notna()]
        left = [n for n in (home, away) if n not in set(res.canon)]
        if len(unres) == 1 and len(left) == 1:
            pairs.at[unres.index[0], "canon"] = left[0]
            changed = True

    for t, c in pairs.loc[pairs.canon.notna() & (pairs.team != pairs.canon), ["team", "canon"]].itertuples(index=False):
        hints.setdefault(t, set()).add(c)
    if not changed:
        break

# สร้างผล win/draw/loss (ลอจิกจากข้อ 5.7 เดิม)
pairs["result"] = np.where(pairs.canon.isna(), None,
                           np.where(pairs.winner.isna(), "draw",
                           np.where(pairs.winner == pairs.canon, "win", "loss")))

key = pairs[["match_id", "team", "canon", "result"]].copy()
key["team"] = key["team"].astype('category')

# Merge ข้อมูล
df_merged_raw = df_all_movements.merge(key, on=["match_id", "team"], how="left", validate="m:1")
df_merged_raw = df_merged_raw[df_merged_raw["canon"].notna()].copy()
df_merged_raw["team"] = df_merged_raw["canon"].astype("category")
df_merged_raw["result"] = df_merged_raw["result"].astype("category")
df_merged_raw = df_merged_raw.drop(columns="canon").reset_index(drop=True)

print(f"\nMerge ข้อมูลเสร็จสิ้น! ได้ DataFrame ที่พร้อมนำไปใช้งาน: df_merged_raw มีขนาด {df_merged_raw.shape}")
```

## 5. Data Cleaning
อธิบายทุกการตัดสินใจ **พร้อมเหตุผล** และรายงานจำนวนแถวที่เปลี่ยนแปลงในแต่ละขั้น

| ขั้นตอน | สิ่งที่ทำ | เหตุผล | แถวคงเหลือ |
|---|---|---|---|
| 5.1 | เก็บเฉพาะ `Pass` + `Carry` (ตัด Shot, Pressure, Duel, Interception, Ball Recovery, Foul Committed, Dribble) | คำถามวิจัยคือ "รูปแบบการเคลื่อนที่" ต้องมีจุดเริ่ม-จุดสิ้นสุดจริง เหตุการณ์อื่นไม่มีปลายทาง (`end_x/end_y` เป็น NaN 25.4%) ซึ่งเป็น structural missing ไม่ใช่ MCAR จึงไม่ impute เพราะจะสร้างการเคลื่อนที่ปลอม ส่วน Shot ปลายทางคือกรอบประตู ไม่ใช่การเคลื่อนที่ของผู้เล่น | 6,844,503 |
| 5.2 | พิกัดนอกสนาม: เกิน ≤ 2 หลา → clip ให้อยู่ใน 0-120 x 0-80, เกิน > 2 หลา → ลบ | พบเฉพาะส่วนน้อยมาก ส่วนใหญ่เป็นความคลาดเคลื่อนเล็กน้อย/บอลออกนอกสนาม จึงคงแถวไว้ด้วยการ clip ส่วนที่เกินมากผิดปกติจึงลบ | *(ดู `cleaning_log_df`)* |
| 5.3 | ลบแถวที่จุดเริ่ม = จุดสิ้นสุด (ระยะ 0) | Carry ระยะ 0 คือไม่ได้เคลื่อนที่ ไม่มีทิศทาง (มุมนิยามไม่ได้) และจะรวมเป็นกลุ่ม "ไม่ขยับ" ที่บิดผล clustering ส่วน Pass ระยะ 0 น่าจะเป็น `end_location` หายแล้วถูกเติมด้วยจุดเริ่มต้นตอนโหลด | *(ดู `cleaning_log_df`)* |
| 5.4 | ลบแถวซ้ำทุกคอลัมน์ (เก็บแถวแรก) | พิกัด 4 ค่า (ละเอียด 0.1 หลา) + ผู้เล่น + แมตช์ เหมือนกันทุกค่า โอกาสบังเอิญต่ำมาก และมีสัดส่วนน้อยมาก (ข้อจำกัด: ตอนโหลดไม่ได้เก็บ event id) | *(ดู `cleaning_log_df`)* |
| 5.5 | ตรวจ outlier ของระยะ (IQR แยกตาม Pass/Carry) แต่ **คงไว้ทั้งหมด** | outlier คือลูกยาว/ลูกโด่ง/การเลี้ยงบอลยาวที่เกิดขึ้นจริง ไม่มีค่าเกินเส้นทแยงสนาม (144.2 หลา) และเป็นลักษณะที่ใช้แยกสไตล์ตรงไปตรงมา ไปจัดการด้วย log-transform / RobustScaler ใน Pipeline แทน | ไม่เปลี่ยน |

>

```python
# =================================================================
# 5. Data Cleaning (บันทึกจำนวนแถวลง cleaning_log)
# =================================================================
import pandas as pd
import numpy as np

cleaning_log = []

def log_step(step, action, n_before, n_after):
    cleaning_log.append({
        "ขั้นตอน": step,
        "สิ่งที่ทำ": action,
        "แถวก่อน": n_before,
        "แถวหลัง": n_after,
        "เปลี่ยนแปลง": n_after - n_before
    })

# ใช้ข้อมูลเริ่มต้นจากข้อ 3 (ที่ merge แล้ว)
df_c = df_merged_raw.copy()
n0 = len(df_c)
print(f"เริ่มต้นข้อ 5: {n0:,} แถว")

COORD = ["start_x", "start_y", "end_x", "end_y"]

# -----------------------------------------------------------------
# 5.1 เก็บเฉพาะ Pass และ Carry
# -----------------------------------------------------------------
n_b = len(df_c)
df_c = df_c[df_c["type"].isin(["Pass", "Carry"])].copy()

# ลดหน่วยความจำ: string -> category, พิกัด -> float32
for c in ["team", "player", "type", "result"]:
    df_c[c] = df_c[c].astype("category")
df_c[COORD] = df_c[COORD].astype("float32")
df_c = df_c.reset_index(drop=True)

log_step("5.1", "เก็บเฉพาะ Pass + Carry", n_b, len(df_c))

# -----------------------------------------------------------------
# 5.2 พิกัดนอกสนาม (สนาม StatsBomb = 120 x 80)
# -----------------------------------------------------------------
n_b = len(df_c)
TOL = 2.0 # หลา

def overshoot(s, hi):
    return np.maximum(-s.to_numpy(), s.to_numpy() - hi).clip(min=0)

over = np.maximum.reduce([
    overshoot(df_c.start_x, 120), overshoot(df_c.end_x, 120),
    overshoot(df_c.start_y, 80), overshoot(df_c.end_y, 80)
])

# กรองเอาเฉพาะแถวที่พิกัดเกินไม่เกิน 2.0 หลา
df_c = df_c[over <= TOL].copy()

# Clip ให้อยู่ในกรอบสนาม
for c in ["start_x", "end_x"]:
    df_c[c] = df_c[c].clip(0, 120)
for c in ["start_y", "end_y"]:
    df_c[c] = df_c[c].clip(0, 80)

df_c = df_c.reset_index(drop=True)
log_step("5.2", f"ลบแถวพิกัดเกินสนาม > {TOL} หลา และ clip ที่เหลือให้อยู่ใน 0-120 x 0-80", n_b, len(df_c))

# -----------------------------------------------------------------
# 5.3 ลบแถวที่จุดเริ่ม = จุดสิ้นสุด (ระยะ 0)
# -----------------------------------------------------------------
n_b = len(df_c)
dist = np.hypot(df_c.end_x - df_c.start_x, df_c.end_y - df_c.start_y)
df_c = df_c[dist > 0].reset_index(drop=True)
log_step("5.3", "ลบแถวที่จุดเริ่ม = จุดสิ้นสุด (ระยะ 0)", n_b, len(df_c))

# -----------------------------------------------------------------
# 5.4 ลบแถวซ้ำทุกคอลัมน์
# -----------------------------------------------------------------
n_b = len(df_c)
df_c = df_c.drop_duplicates().reset_index(drop=True)
log_step("5.4", "ลบแถวซ้ำทุกคอลัมน์ (เก็บแถวแรก)", n_b, len(df_c))

# -----------------------------------------------------------------
# 5.5 ตรวจ Outlier ของระยะเคลื่อนที่ (IQR) -> รายงานแต่ไม่ลบ
# -----------------------------------------------------------------
dist = np.hypot(df_c.end_x - df_c.start_x, df_c.end_y - df_c.start_y)
g = dist.groupby(df_c["type"], observed=True)
q = g.quantile([0.25, 0.75]).unstack()
upper_bound = q[0.75] + 1.5 * (q[0.75] - q[0.25])
is_out = dist > df_c["type"].astype(str).map(upper_bound).astype(float)

rep = pd.DataFrame({
    "Q1": q[0.25],
    "Q3": q[0.75],
    "ขอบบน IQR": upper_bound,
    "จำนวน outlier": is_out.groupby(df_c["type"], observed=True).sum(),
    "สัดส่วน": is_out.groupby(df_c["type"], observed=True).mean()
}).round(3)

print("\nOutlier ของระยะเคลื่อนที่ (IQR แยกตามชนิด): ")
print(rep.to_string())
print(f"ระยะสูงสุด {dist.max():.1f} หลา (เส้นทแยงสนาม = 144.2) -> ไม่มีค่าเป็นไปไม่ได้")
log_step("5.5", "ตรวจ outlier ระยะ (IQR) แต่คงไว้ทั้งหมด", len(df_c), len(df_c))



df_clean = df_c.copy()

# =================================================================
# สรุปผล
# =================================================================
print("\n=== สรุปผลการ Data Cleaning ===")
cleaning_log_df = pd.DataFrame(cleaning_log)
display(cleaning_log_df)

print(f"\nข้อมูลสะอาด df_clean: {df_clean.shape[0]:,} แถว x {df_clean.shape[1]} คอลัมน์")
print(f"(เหลือ {len(df_clean)/n0*100:.1f}% จากข้อมูลก่อน Clean) | หน่วยความจำที่ใช้: {df_clean.memory_usage(deep=True).sum()/1e6:,.1f} MB")
display(df_clean.head())
```

**Pass (การส่งบอล)**
* ความหมาย: คือการเตะหรือส่งลูกฟุตบอลจากผู้เล่นคนหนึ่งไปยังเพื่อนร่วมทีม เพื่อรักษาการครองบอล เปลี่ยนพื้นที่เล่น หรือเจาะแนวรับ
* ลักษณะการใช้งาน: มีหลายรูปแบบ เช่น การส่งสั้น (Short pass), การวางบอลยาว (Long pass), การส่งแบบสัมผัสเดียว (One-touch pass), หรือการจ่ายบอลทะลุช่องเพื่อทำลายแนวรับ (Killer Pass)
* เป้าหมาย: ใช้ความเร็วของลูกบอลที่เคลื่อนที่ผ่านอากาศหรือเรียดไปกับพื้น เพื่อส่งต่อความได้เปรียบให้เพื่อนเล่นต่อได้ทันที


**Carry (การพาบอล)**
* ความหมาย: คือการที่ผู้เล่นครอบครองลูกบอลแล้ว "พาบอลด้วยเท้าตัวเอง" วิ่งเคลื่อนที่ขึ้นไปข้างหน้าผ่านพื้นที่ว่าง
* ข้อแตกต่างจาก Dribble (การเลี้ยงหลบ):

	• Dribble เน้นการเลี้ยงหลบ 1-ต่อ-1 เพื่อผ่านตัวประกบหรือคู่ต่อสู้โดยตรง

	• Carry เน้นการ "ลำเลียงบอล" หรือพาบอลรุกคืบเข้าไปในแดนคู่ต่อสู้เพื่อกินพื้นที่ (ไม่จำเป็นต้องมีการเผชิญหน้าหรือหลบผู้เล่นฝ่ายตรงข้ามเสมอไป)
* Progressive Carry (การพาบอลก้าวหน้า): ในสถิติฟุตบอลยุคใหม่ หมายถึงการพาบอลขึ้นหน้าอย่างมีความหมาย (เช่น ระยะทางอย่างน้อย 10 หลาในแดนทั่วไป หรือ 5 หลาเมื่อเข้าใกล้กรอบเขตโทษ) เพื่อสร้างจังหวะรุกให้ทีม



```python

# Pitch heatmaps for Pass/Carry start and end locations
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle, Arc

# 1. ฟังก์ชันวาดเส้นสนามฟุตบอล
def draw_pitch(ax):
    # ขอบสนาม
    line_color = "#BDBDBD"
    lw = 2

    ax.add_patch(Rectangle(
        (0, 0), 120, 80,
        fill=False, edgecolor=line_color, linewidth=lw, zorder=3
    ))

    # เส้นแบ่งครึ่งสนาม
    ax.plot([60, 60], [0, 80],
            color=line_color, linewidth=lw, zorder=3)

    # วงกลมกลางสนาม
    ax.add_patch(Circle(
        (60, 40), 10,
        fill=False, edgecolor=line_color, linewidth=lw, zorder=3
    ))

    # จุดกลางสนาม
    ax.scatter([60], [40], color=line_color, s=8, zorder=4)

    # เขตโทษทั้งสองฝั่ง
    ax.add_patch(Rectangle(
        (0, 18), 18, 44,
        fill=False, edgecolor=line_color, linewidth=lw, zorder=3
    ))
    ax.add_patch(Rectangle(
        (102, 18), 18, 44,
        fill=False, edgecolor=line_color, linewidth=lw, zorder=3
    ))

    # เขตประตูทั้งสองฝั่ง
    ax.add_patch(Rectangle(
        (0, 30), 6, 20,
        fill=False, edgecolor=line_color, linewidth=lw, zorder=3
    ))
    ax.add_patch(Rectangle(
        (114, 30), 6, 20,
        fill=False, edgecolor=line_color, linewidth=lw, zorder=3
    ))

    # จุดยิงจุดโทษ
    ax.scatter([12, 108], [40, 40],
               color=line_color, s=8, zorder=4)

    # ส่วนโค้งหน้ากรอบเขตโทษ
    ax.add_patch(Arc(
        (12, 40), 20, 20,
        theta1=307, theta2=53,
        color=line_color, linewidth=lw, zorder=3
    ))
    ax.add_patch(Arc(
        (108, 40), 20, 20,
        theta1=127, theta2=233,
        color=line_color, linewidth=lw, zorder=3
    ))

    # กำหนดขอบเขตและสัดส่วนสนาม
    ax.set_xlim(0, 120)
    ax.set_ylim(0, 80)
    ax.set_aspect("equal", adjustable="box")


# 2. เตรียมข้อมูลสำหรับวาด Heatmap
plot_df = eda[
    eda["type"].isin(["Pass", "Carry"])
    & eda[["start_x", "start_y", "end_x", "end_y"]]
        .notna().all(axis=1)
].copy()

plot_df = plot_df.sample(
    n=min(200_000, len(plot_df)),
    random_state=425
)

# 3. วาด Heatmap ทั้ง 4 รูป
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

plot_specs = [
    ("Pass", "start_x", "start_y", axes[0, 0],
     "Pass - Start Location"),
    ("Pass", "end_x", "end_y", axes[0, 1],
     "Pass - End Location"),
    ("Carry", "start_x", "start_y", axes[1, 0],
     "Carry - Start Location"),
    ("Carry", "end_x", "end_y", axes[1, 1],
     "Carry - End Location")
]

for event_type, x_col, y_col, ax, title in plot_specs:
    subset = plot_df[plot_df["type"] == event_type]

    if subset.empty:
        ax.set_title(f"{title} (No data)")
        continue

    # Heatmap
    hist = ax.hist2d(
        subset[x_col],
        subset[y_col],
        bins=[24, 16],
        range=[[0, 120], [0, 80]],
        cmap="YlOrRd",
        cmin=1,
        zorder=1
    )

    # วาดเส้นสนามทับลงบน Heatmap
    draw_pitch(ax)

    ax.set_xlabel("Pitch X")
    ax.set_ylabel("Pitch Y")
    ax.set_title(title)

    fig.colorbar(
        hist[3],
        ax=ax,
        label="Number of Events",
        fraction=0.046,
        pad=0.03
    )

plt.suptitle(
    "Spatial Distribution of Pass and Carry",
    fontsize=15
)
plt.tight_layout()
plt.show()
```

**1. รูปแบบการส่งบอล (Pass)**  

- จุดเริ่มต้น (Start Location): มีความหนาแน่นสูงมาก (สีแดงเข้ม) ที่จุดตั้งเตะตามกฎการเล่น เช่น บริเวณหน้าประตูตัวเอง ($X \approx 10, Y \approx 40$) จากลูกตั้งเตะจากประตู (Goal Kick) และบริเวณจุดกลางสนาม ($X \approx 60, Y \approx 40$) จากการเขี่ยบอลเริ่มเล่น   

- จุดสิ้นสุด (End Location): ปลายทางการส่งบอลกระจายตัวกว้างทั่วสนาม แต่จะหนาแน่นเป็นพิเศษบริเวณริมเส้นทั้งสองฝั่ง และจุดยิงประตูในกรอบเขตโทษของคู่แข่ง ($X \approx 108, Y \approx 40$) แสดงให้เห็นว่าการจ่ายบอลถูกใช้เป็นเครื่องมือหลักในการเจาะเข้าพื้นที่อันตราย   

**2. รูปแบบการพาบอล/เลี้ยงบอล (Carry)**  

- จุดเริ่มต้นและสิ้นสุด (Start & End Location): ทั้งจุดเริ่มและจุดจบของการ Carry กระจุกตัวอยู่อย่างชัดเจนบริเวณ ช่องริมเส้นทั้งฝั่งซ้ายและฝั่งขวา ($Y \approx 0-15$ และ $Y \approx 65-80$) ตลอดแนวแดนกลางไปจนถึงแดนหน้า ($X \approx 40-90$)   

- ลักษณะการใช้งาน: การพาบอลถูกใช้เป็นกลยุทธ์หลักในการดันเกมขึ้นหน้าตามริมเส้น (Wing Progression) มากกว่าการเลี้ยงบอลตัดเข้าพื้นที่กลางสนาม  

**3. ข้อสรุปเชิงแท็กติก (Tactical Insight)**

- Pass (ส่งบอล): มีบทบาทในการสร้างสรรค์เกมจากแนวหลัง การเปลี่ยนแกนการเล่น และการจ่ายบอลจังหวะสุดท้ายเข้ากรอบเขตโทษเพื่อทำประตู  

- Carry (พาบอล): มีบทบาทในการครอบครองบอลเพื่อกินพื้นที่ตามแนวยาวริมเส้น ช่วยให้ทีมดันบล็อกเกมรุกขึ้นสูงก่อนที่จะเลือกจ่ายบอลต่อเข้าสู่พื้นที่เข้าทำ

```python
# ==========================================
# Target Variable Analysis (Target vs Features)
# ==========================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------
# Part 1: Event-Level Analysis
# ------------------------------------------
print("=== 1. Target Distribution (Match Results - Event Level) ===")
result_share = eda["result"].value_counts(normalize=True).to_frame("Proportion (%)") * 100
display(result_share.round(2))

# สร้างตัวแปรแบ่งกลุ่ม Win vs Non-Win
eda["match_outcome"] = np.where(eda["result"] == "win", "Win Matches", "Non-Win Matches")


# ------------------------------------------
# Part 2: Team-Match Level Analysis
# ------------------------------------------
print("\n=== 2. Team-Match Level Tactical Metrics Summary ===")

# ใช้เฉพาะ Pass และ Carry
tc = eda[eda["type"].isin(["Pass", "Carry"])].copy()
tc["_result"] = tc["result"].astype("string").str.strip().str.lower()

# คำนวณคุณลักษณะของแต่ละทีมในแต่ละแมตช์
team_match = (
    tc.groupby(["match_id", "team", "_result"], observed=True)
    .agg(
        Total_Events=("type", "size"),
        Pass_Count=("type", lambda s: (s == "Pass").sum()),
        Carry_Count=("type", lambda s: (s == "Carry").sum())
    )
    .reset_index()
)

pass_distance = (
    tc[tc["type"] == "Pass"]
    .groupby(["match_id", "team", "_result"], observed=True)["dist"]
    .mean()
    .rename("Mean_Pass_Distance")
    .reset_index()
)

carry_distance = (
    tc[tc["type"] == "Carry"]
    .groupby(["match_id", "team", "_result"], observed=True)["dist"]
    .mean()
    .rename("Mean_Carry_Distance")
    .reset_index()
)

team_match = team_match.merge(
    pass_distance, on=["match_id", "team", "_result"], how="left"
).merge(
    carry_distance, on=["match_id", "team", "_result"], how="left"
)

team_match["Pass_Pct"] = (team_match["Pass_Count"] / team_match["Total_Events"] * 100)
team_match["Carry_Pct"] = (team_match["Carry_Count"] / team_match["Total_Events"] * 100)

valid_results = ["win", "draw", "loss"]
team_match_valid = team_match[team_match["_result"].isin(valid_results)].copy()

if not team_match_valid.empty:
    result_table = (
        team_match_valid.groupby("_result", observed=True)
        .agg(
            Team_Match_Count=("match_id", "size"),
            Mean_Pass_Pct=("Pass_Pct", "mean"),
            Mean_Carry_Pct=("Carry_Pct", "mean"),
            Mean_Pass_Distance=("Mean_Pass_Distance", "mean"),
            Mean_Carry_Distance=("Mean_Carry_Distance", "mean")
        )
        .reindex(["win", "draw", "loss"])
        .rename_axis("Result")
        .reset_index()
    )

    result_table = result_table.rename(columns={
        "Team_Match_Count": "จำนวนทีม-แมตช์",
        "Mean_Pass_Pct": "Pass เฉลี่ย (%)",
        "Mean_Carry_Pct": "Carry เฉลี่ย (%)",
        "Mean_Pass_Distance": "ระยะ Pass เฉลี่ย",
        "Mean_Carry_Distance": "ระยะ Carry เฉลี่ย"
    })

    numeric_cols = result_table.select_dtypes(include="number").columns
    result_table[numeric_cols] = result_table[numeric_cols].round(2)

    display(result_table)
```

**ผลต่อการเตรียมข้อมูล (Data Preparation)**

1. การทำ Feature Scaling: ต้องใช้ StandardScaler หรือ MinMaxScaler ปรับพิกัด X, Y และระยะทางให้อยู่ในสเกลมาตรฐาน เพื่อป้องกันไม่ให้ฟีเจอร์ที่มีช่วงตัวเลขกว้างกว่าครอบงำการคำนวณระยะห่างในโมเดล Clustering

2. Spatial Feature Engineering: ต้องใช้พิกัดจุดเริ่มต้น $(start\_x, start\_y)$, จุดสิ้นสุด $(end\_x, end\_y)$, และทิศทาง $(\Delta X, \Delta Y)$ ร่วมกัน เพราะระยะทางหรือมุมโดดๆ ไม่สามารถแยกพฤติกรรมได้   

**ผลต่อการเลือกโมเดลและการสรุปผล (Model Strategy)**

1. เลือกใช้ Unsupervised Spatial Clustering:
- ในเมื่อสถิติภาพรวมแยกทีมชนะไม่ได้ จึงจำเป็นต้องใช้ Spatial Clustering (เช่น MiniBatchKMeans หรือ GMM) เพื่อเปลี่ยนพิกัดดิบให้กลายเป็น Movement Archetypes (กลุ่มรูปแบบแท็กติกเชิงพื้นที่)   

2. แปลงข้อมูลสู่ Team-Match Level Aggregation:

- นำ Movement Archetypes ที่สกัดได้ มานับสัดส่วนสะสมการเกิดจริงของแต่ละทีมในแต่ละแมตช์



## 7. Feature Engineering
| Feature | สร้างจาก | เหตุผล |
|---|---|---|
| `start_x` | `start_x` (ใช้ตรง ๆ) | แสดงตำแหน่งเริ่มต้นตามแนวยาวของสนาม ช่วยแยกการเคลื่อนที่ในแดนรับ แดนกลาง และแดนบุก |
| `start_y` | `start_y` (ใช้ตรง ๆ) | แสดงตำแหน่งเริ่มต้นตามแนวกว้างของสนาม ช่วยแยกการเล่นบริเวณริมเส้นและบริเวณกลางสนาม |
| `log_dist` | `log1p(hypot(end_x-start_x, end_y-start_y))` | ระยะทางมีการกระจายแบบเบ้ขวา การแปลงด้วย log ช่วยลดอิทธิพลของการเคลื่อนที่ระยะไกลและค่าผิดปกติต่อ K-Means |
| `angle_sin` | `(end_y-start_y) / dist` | ใช้แทนองค์ประกอบแนวตั้งของทิศทางการเคลื่อนที่ โดยแปลงมุมเป็นค่า sine เพื่อให้แทนทิศทางได้เหมาะสม |
| `angle_cos` | `(end_x-start_x) / dist` | ใช้คู่กับ angle_sin เพื่อแทนองค์ประกอบแนวนอนของทิศทางการเคลื่อนที่ ค่าบวกหมายถึงเคลื่อนที่ไปทางแกน X บวก และค่าลบหมายถึงเคลื่อนที่ไปทางแกน X ลบ |
| `is_pass` | `type == "Pass"` (Pass = 1, Carry = 0) | ช่วยให้โมเดลแยกประเภทการเคลื่อนที่ระหว่างการส่งบอลกับการพาบอล ซึ่งอาจมีรูปแบบและระยะทางแตกต่างกัน |
| `goal_progress` | ระยะจากจุดเริ่มต้นถึงประตู `(120, 40)` − ระยะจากจุดสิ้นสุดถึงประตู `(120, 40)` | วัดการเปลี่ยนแปลงของระยะห่างจากจุดอ้างอิง (120, 40) ค่าบวกหมายถึงจุดสิ้นสุดอยู่ใกล้จุดอ้างอิงกว่าจุดเริ่มต้น ส่วนค่าลบหมายถึงจุดสิ้นสุดอยู่ไกลขึ้น ทั้งนี้ การตีความว่าเข้าใกล้ประตูคู่แข่งต้องตรวจสอบระบบพิกัดของข้อมูลต้นทาง|

**ตัวแปรที่ไม่นำมาใช้เป็น Feature**

- `end_x`, `end_y`, `dx`, `dy` — ไม่นำมาใช้โดยตรง เพราะใช้คำนวณ Feature ด้านระยะทางและทิศทางแล้ว
- `start_third`, `end_third`, `direction` — เป็นตัวแปรที่จัดกลุ่มหรือสรุปจากตำแหน่งและทิศทาง จึงเก็บไว้ช่วยตีความผล
- `result`, `team`, `player`, `match_id` — เก็บไว้ใช้วิเคราะห์และตีความผลการจัดกลุ่ม ไม่ใช้เป็น Input ของโมเดล

**การตรวจสอบข้อมูลก่อนสร้างโมเดล**

ตรวจสอบว่า Feature ไม่มีค่า `NaN` หรือ `inf` และปรับสเกลด้วย `StandardScaler` ก่อนใช้ K-Means เพื่อให้ตัวแปรที่มีช่วงค่าต่างกันไม่ส่งผลต่อการคำนวณระยะห่างมากเกินไป

**ข้อควรระวัง:** Feature ที่ใช้ `end_x` และ `end_y` จะทราบค่าได้หลังเหตุการณ์สิ้นสุดแล้ว ดังนั้นชุด Feature นี้เหมาะกับการจัดกลุ่มรูปแบบการเคลื่อนที่ที่เกิดขึ้นแล้ว หากต้องการทำนายก่อนเหตุการณ์เกิดขึ้น ต้องตัด Feature ที่ต้องอาศัยจุดสิ้นสุดออก

```python
# ==========================================================
# 7. Feature Engineering
# ==========================================================
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GroupShuffleSplit

GOAL_X, GOAL_Y = 120.0, 40.0

# ใช้ข้อมูลหลัง Cleaning โดยไม่แก้ไข df_clean ต้นฉบับ
src = df_clean

sx = src["start_x"].to_numpy(dtype="float32")
sy = src["start_y"].to_numpy(dtype="float32")
ex = src["end_x"].to_numpy(dtype="float32")
ey = src["end_y"].to_numpy(dtype="float32")

# คำนวณการเปลี่ยนตำแหน่ง ระยะทาง และระยะห่างจากประตู
dx = ex - sx
dy = ey - sy
dist = np.hypot(dx, dy)

assert (dist > 0).all(), "พบระยะทาง 0 กรุณาตรวจสอบขั้นตอน Cleaning"

goal_start = np.hypot(GOAL_X - sx, GOAL_Y - sy)
goal_end = np.hypot(GOAL_X - ex, GOAL_Y - ey)

# ----------------------------------------------------------
# 1) สร้าง Features สำหรับ Clustering
# ----------------------------------------------------------
fe = pd.DataFrame({
    "start_x": sx,
    "start_y": sy,
    "log_dist": np.log1p(dist),
    "angle_sin": dy / dist,
    "angle_cos": dx / dist,
    "is_pass": (src["type"] == "Pass").to_numpy().astype("float32"),
    "goal_progress": goal_start - goal_end,
}).astype("float32")

FEATURES = list(fe.columns)

print("Features ที่ใช้:", FEATURES)
print(f"ขนาดข้อมูล: {fe.shape}")
print(f"หน่วยความจำ: {fe.memory_usage(deep=True).sum()/1e6:,.1f} MB")
display(fe.describe().T.round(3))

# ----------------------------------------------------------
# 2) เก็บตัวแปรสำหรับตีความผลภายหลัง
# ไม่ใช้เป็น Input ของ Clustering
# ----------------------------------------------------------
meta = src[["match_id", "team", "player", "type", "result"]].copy()
meta["dx"] = dx
meta["dy"] = dy
meta["dist"] = dist

# ----------------------------------------------------------
# 3) ตรวจสอบ Feature ที่ไม่ควรเข้าโมเดล
# ----------------------------------------------------------
FORBIDDEN = {"result", "team", "player", "match_id", "winner"}

assert FORBIDDEN.isdisjoint(FEATURES), \
    "พบตัวแปรผลการแข่งขันหรือตัวระบุหลุดเข้า Features"

# ตรวจสอบ NaN, inf และช่วงค่าของมุม
assert np.isfinite(fe.to_numpy()).all(), "พบ NaN หรือ inf"

assert fe["angle_sin"].between(-1, 1).all()
assert fe["angle_cos"].between(-1, 1).all()

assert len(fe) == len(src), "จำนวนแถวไม่ตรงกับข้อมูลต้นฉบับ"

print("ผ่านการตรวจสอบ Feature เบื้องต้น")

# ----------------------------------------------------------
# 4) ตรวจสอบว่า Feature แต่ละตัวรู้ค่าเมื่อใด
# ----------------------------------------------------------
timing = pd.DataFrame({
    "feature": FEATURES,
    "รู้ค่าเมื่อ": [
        "จุดเริ่ม",
        "จุดเริ่ม",
        "จุดจบ",
        "จุดจบ",
        "จุดจบ",
        "เมื่อทราบประเภท Event",
        "จุดจบ"
    ],
    "หมายเหตุ": [
        "พิกัดเริ่มต้นแกน X",
        "พิกัดเริ่มต้นแกน Y",
        "คำนวณจากพิกัดเริ่มต้นและปลายทาง",
        "คำนวณจากการเปลี่ยนแปลงตำแหน่ง",
        "คำนวณจากการเปลี่ยนแปลงตำแหน่ง",
        "Pass = 1, Carry = 0",
        "คำนวณระยะห่างจากจุดอ้างอิง (120, 40)"
    ]
})

display(timing)

# ----------------------------------------------------------
# 5) ตรวจสอบความสัมพันธ์ระหว่าง Features
# ใช้ตัวอย่าง 200,000 แถวเพื่อลดการใช้หน่วยความจำ
# ----------------------------------------------------------
corr = fe.sample(
    n=min(200_000, len(fe)),
    random_state=42
).corr().abs()

upper = corr.where(
    np.triu(np.ones(corr.shape, dtype=bool), k=1)
)

pairs = (
    upper.stack()
    .loc[lambda s: s > 0.7]
    .sort_values(ascending=False)
)

print("คู่ Feature ที่มี |correlation| > 0.7:")
print(pairs.round(3).to_string() if len(pairs) else "ไม่พบ")
```

**ตรวจ leakage:** งานนี้เป็น unsupervised ที่จัดกลุ่ม "การเคลื่อนที่ที่เกิดขึ้นแล้ว" (descriptive)
ดังนั้นเวลาที่ใช้ feature คือ ณ จุดที่การเคลื่อนที่จบ feature ทุกตัวคำนวณจากพิกัดของแถวนั้นเอง
(start/end) ไม่ใช้ผลแมตช์ (`result`), ชื่อทีม, ชื่อผู้เล่น หรือสถิติรวมข้ามแถว
`result`/`team` ใช้หลัง clustering เพื่อดู archetype mix ของทีมที่ชนะบ่อยเท่านั้น
feature ที่ใช้ข้อมูลจุดจบ (log_dist, angle, goal_progress) จะใช้ไม่ได้ ถ้าภายหลังเปลี่ยนโจทย์
เป็น "ทำนายก่อนส่ง/พาบอล"

## 8. Data Partitioning
อธิบายวิธีแบ่ง train / validation / test และเหตุผลที่วิธีนี้เหมาะกับข้อมูลและคำถามของกลุ่ม

**วิธีการแบ่งข้อมูล (Data Partitioning Method)**

* ใช้เทคนิค Group-based Split ผ่านโมดูล `GroupShuffleSplit` โดยกำหนดให้ `match_id` เป็นกลุ่ม (Group) ในการแบ่งชุดข้อมูล


* แบ่งข้อมูลออกเป็น Train 70%, Validation 15% และ Test 15% โดยสุ่มแยกที่ระดับ "แมตช์การแข่งขัน" แทนการสุ่มแยกทีละแถวเหตุการณ์ (Random Split)


* กระบวนการนี้รับประกันว่าเหตุการณ์การเคลื่อนที่ (Pass และ Carry) ทั้งหมดที่เกิดขึ้นในแมตช์เดียวกัน จะถูกจัดรวมไว้ในชุดข้อมูลใดชุดข้อมูลหนึ่งเพียงชุดเดียว ทำให้ไม่มีข้อมูลแมตช์ที่ซ้ำซ้อนกันระหว่างชุด Train, Validation และ Test



**เหตุผลที่วิธีนี้เหมาะกับข้อมูลและคำถาม**

* **การป้องกันข้อมูลรั่วไหล (Data Leakage):** ข้อมูลเหตุการณ์ฟุตบอลมีความสัมพันธ์และต่อเนื่องกันในแต่ละเกม การสุ่มแยกทีละแถวจะทำให้เหตุการณ์ที่ต่อเนื่องกันถูกแยกไปอยู่ทั้งชุด Train และ Test ส่งผลให้โมเดลจดจำรูปแบบเฉพาะของแมตช์นั้นและเกิด Overfitting การแบ่งด้วย `match_id` จึงช่วยป้องกันการรั่วไหลของข้อมูลภายในแมตช์ได้อย่างเด็ดขาด


* **การตอบโจทย์ด้านยุทธวิธี (Tactical Insights):** คำถามวิจัยของกลุ่มคือการศึกษา "รูปแบบการเคลื่อนที่ทำทาง (Movement Archetypes)" และสไตล์การเล่นของทีมที่มีอัตราการชนะสูง แท็กติกของทีมฟุตบอลจะเปลี่ยนไปตามแต่ละแมตช์ การทดสอบโมเดลด้วยแมตช์ที่โมเดลไม่เคยเห็นมาก่อนเลยทั้งเกม จะช่วยพิสูจน์ว่ารูปแบบการบุกที่เป็นแม่แบบสากลที่ค้นพบนั้นใช้งานได้จริงและสะท้อนโครงสร้างเกมบุกได้ตรงจุด


* **การประเมินความเสถียรของโมเดล Unsupervised Learning:** การจัดกลุ่มเพื่อสร้างตัวแปรเป้าหมาย (Target) หรือ Cluster ID ต้องการความเสถียร การแบ่งชุด Validation และ Test แยกออกมาเป็นแมตช์ใหม่ จะช่วยยืนยันว่าจุดศูนย์กลางของกลุ่ม (Centroids) ครอบคลุมรูปแบบการเคลื่อนที่ทั่วไป ไม่ได้เกิดจากการเรียนรู้จำเพาะกับชุดข้อมูล Train เพียงอย่างเดียว

```python
# ==========================================
# 8. Data Partitioning (Train 70% / Validation 15% / Test 15%)
# ==========================================
from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import StandardScaler

# 1. แบ่งรอบแรก: Train (70%) และ ชุดชั่วคราว Temp (30%)
splitter_train = GroupShuffleSplit(n_splits=1, test_size=0.30, random_state=42)
train_idx, temp_idx = next(
    splitter_train.split(fe[FEATURES], groups=meta["match_id"])
)

# 2. แบ่งรอบสอง: นำชุดชั่วคราว Temp (30%) มาแบ่งครึ่งเป็น Validation (15%) และ Test (15%)
temp_groups = meta.iloc[temp_idx]["match_id"]
splitter_val = GroupShuffleSplit(n_splits=1, test_size=0.50, random_state=42)

# ผลลัพธ์จะเป็น index สัมพัทธ์ของชุด temp

val_rel_idx, test_rel_idx = next(
    splitter_val.split(fe.iloc[temp_idx], groups=temp_groups)
)

# แปลงค่า index สัมพัทธ์กลับไปเป็น index ของ DataFrame ต้นฉบับ
val_idx = temp_idx[val_rel_idx]
test_idx = temp_idx[test_rel_idx]

# ==========================================
# รายงานผลการแบ่งข้อมูล
# ==========================================
print(f"จำนวนเหตุการณ์ใน Train: {len(train_idx):,} แถว (แมตช์ที่ไม่ซ้ำ: {meta.iloc[train_idx]['match_id'].nunique():,})")
print(f"จำนวนเหตุการณ์ใน Validation: {len(val_idx):,} แถว (แมตช์ที่ไม่ซ้ำ: {meta.iloc[val_idx]['match_id'].nunique():,})")
print(f"จำนวนเหตุการณ์ใน Test: {len(test_idx):,} แถว (แมตช์ที่ไม่ซ้ำ: {meta.iloc[test_idx]['match_id'].nunique():,})")

# ตรวจสอบ Data Leakage (ต้องเป็น 0 ทั้งหมด)
train_matches = set(meta.iloc[train_idx]["match_id"])
val_matches = set(meta.iloc[val_idx]["match_id"])
test_matches = set(meta.iloc[test_idx]["match_id"])

print("\nตรวจสอบ Data Leakage (จำนวนแมตช์ที่ซ้ำซ้อนกันข้ามชุดข้อมูล):")
print(f"Train vs Validation: {len(train_matches & val_matches)} แมตช์")
print(f"Train vs Test: {len(train_matches & test_matches)} แมตช์")
print(f"Validation vs Test: {len(val_matches & test_matches)} แมตช์")
```

## 9. Preprocessing Pipeline
การจัดการ missing values, encoding, scaling ที่ต้อง "เรียนรู้จากข้อมูล" ให้ fit จาก **train set เท่านั้น**
แล้วนำไป transform ชุดอื่น (แนะนำใช้ `sklearn.pipeline.Pipeline` / `ColumnTransformer`)

```python
# ==========================================
# 9. Preprocessing Pipeline
# ==========================================
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
import pandas as pd

# 1. กำหนดกลุ่ม Features จากข้อมูลที่สร้างไว้ใน Feature Engineering
num_features = ['start_x', 'start_y', 'log_dist', 'angle_sin', 'angle_cos', 'goal_progress']
bin_features = ['is_pass']

# 2. สร้าง Pipeline สำหรับข้อมูลเชิงตัวเลข (Continuous Features)
# จัดการ Missing Values + ปรับสเกลข้อมูล (Standardization)
num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),  # รองรับกรณีมีค่า NaN หลุดเข้ามา
    ('scaler', StandardScaler())                     # ปรับสเกลด้วย Mean=0, SD=1
])

# 3. สร้าง Pipeline สำหรับข้อมูล Binary
bin_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')) # กรณีมีค่า Missing ให้แทนด้วยค่าที่พบบ่อยสุด
])

# 4. รวม Pipeline ด้วย ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', num_pipeline, num_features),
        ('bin', bin_pipeline, bin_features)
    ],
    remainder='drop'
)

# ==========================================
# การ Fit จาก Train Set เท่านั้น แล้ว Transform ชุดอื่น
# ==========================================

# แยกชุดข้อมูล Feature ตาม Index จากขั้นตอน Data Partitioning
X_train = fe.iloc[train_idx][FEATURES]
X_val = fe.iloc[val_idx][FEATURES]
X_test = fe.iloc[test_idx][FEATURES]

# Fit และ Transform เฉพาะ Train Set
X_train_preprocessed = preprocessor.fit_transform(X_train)

# Transform ชุด Validation และ Test โดยใช้ค่าสถิติ (Mean, SD) ที่เรียนรู้จาก Train Set เท่านั้น
X_val_preprocessed = preprocessor.transform(X_val)
X_test_preprocessed = preprocessor.transform(X_test)

# รายงานผลลัพธ์
print("Preprocessing Pipeline ดำเนินการสำเร็จ!")
print(f"ขนาดข้อมูล X_train_preprocessed: {X_train_preprocessed.shape}")
print(f"ขนาดข้อมูล X_val_preprocessed:   {X_val_preprocessed.shape}")
print(f"ขนาดข้อมูล X_test_preprocessed:  {X_test_preprocessed.shape}")
```

## 10. Model Selection and Hyperparameter Tuning
* เลือกโมเดลอย่างน้อย 3 แบบ และอธิบายเหตุผลที่เลือก
* tuning ด้วย cross-validation หรือ validation set (ห้ามใช้ test set)
* เลือก metric ที่เหมาะกับปัญหา และอธิบายเหตุผล

เลือกโมเดลในกลุ่ม Unsupervised Spatial Clustering จำนวน 3 แบบ เพื่อเปรียบเทียบประสิทธิภาพในการสกัด Movement Archetypes:

1. MiniBatchKMeans (Primary Baseline Model):

- เหตุผลที่เลือก: เหมาะสมที่สุดสำหรับข้อมูลขนาดใหญ่ (Large-Scale Dataset) ระดับล้านแถว เนื่องจากใช้กระบวนการสุ่มอ่านข้อมูลทีละ Mini-batch ช่วยลดการใช้หน่วยความจำ (RAM) และเพิ่มความเร็วในการคำนวณได้อย่างมหาศาลเมื่อเทียบกับ KMeans ดั้งเดิม

2. KMeans (Standard Baseline Model):

- เหตุผลที่เลือก: ใช้เป็นโมเดลมาตรฐานเพื่อเปรียบเทียบ (Benchmark) คุณภาพของจุดศูนย์กลางกลุ่ม (Centroids) และค่าความถูกต้องทางคณิตศาสตร์กับ MiniBatchKMeans ว่าการใช้ Mini-batch ทำให้คุณภาพของการจัดกลุ่มสูญเสียไปอย่างมีนัยสำคัญหรือไม่

3. Gaussian Mixture Models - GMM (Probabilistic Model):

- เหตุผลที่เลือก: เป็นการจัดกลุ่มแบบ Soft Clustering ที่สมมติฐานว่าข้อมูลมาจากกลุ่มการกระจายตัวแบบ Gaussian หลายชุด เหมาะกับการจับรูปแบบพฤติกรรมเชิงพื้นที่ที่ขอบเขตของแต่ละกลุ่มซ้อนทับกัน (Soft Boundaries) เช่น ทิศทางและระยะการจ่ายบอลที่มีความยืดหยุ่นสูง

```python
# ============================================================
# SECTION 10: MODEL SELECTION AND HYPERPARAMETER TUNING
# ============================================================

import time
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import MiniBatchKMeans, KMeans
from sklearn.mixture import GaussianMixture
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score,
    adjusted_rand_score
)

# 1. ตรวจสอบข้อมูลจากข้อ 8-9
if "X_train_preprocessed" not in globals() or "X_val_preprocessed" not in globals():
    raise NameError("ไม่พบ X_train_preprocessed กรุณารันเซลล์ Preprocessing (ข้อ 8-9) ก่อน")

Xtr_all = np.asarray(X_train_preprocessed, dtype=np.float32)
Xva_all = np.asarray(X_val_preprocessed, dtype=np.float32)

print(f"เตรียมข้อมูลเรียบร้อย -> Train: {len(Xtr_all):,} แถว | Val: {len(Xva_all):,} แถว")

# 2. สุ่มข้อมูลสำหรับการจูน เพื่อลดเวลาและ RAM
rng = np.random.default_rng(42)

n_train = min(300_000, len(Xtr_all))
n_val = min(100_000, len(Xva_all))

train_idx_sample = rng.choice(len(Xtr_all), n_train, replace=False)
val_idx_sample = rng.choice(len(Xva_all), n_val, replace=False)

Xtr = Xtr_all[train_idx_sample]
Xva = Xva_all[val_idx_sample]

print(f"Train ที่ใช้จูน: {len(Xtr):,} แถว | Validation ที่ใช้ประเมิน: {len(Xva):,} แถว")
```

```python
# 3. ฟังก์ชันประเมินผล
def evaluate(X, labels):
    labels = np.asarray(labels)
    n_clusters = len(np.unique(labels))

    if n_clusters < 2 or n_clusters >= len(labels):
        return {
            "silhouette": np.nan,
            "davies_bouldin": np.nan,
            "calinski_harabasz": np.nan
        }

    sample_size = min(20_000, len(X))
    return {
        "silhouette": silhouette_score(X, labels, sample_size=sample_size, random_state=42),
        "davies_bouldin": davies_bouldin_score(X, labels),
        "calinski_harabasz": calinski_harabasz_score(X, labels)
    }

# 4. คำนวณ Baseline (เทียบกับการแยกแค่ Pass/Carry และการสุ่ม)
is_pass_idx = 6 # ตำแหน่ง Index ของ is_pass (หลังจาก 6 num_features)
base = {
    "baseline: Pass/Carry (k=2)": Xva[:, is_pass_idx].astype(int),
    "baseline: Random labels (k=5)": rng.integers(0, 5, size=len(Xva))
}

baseline_rows = []
for name, labels in base.items():
    baseline_rows.append({"baseline": name, **evaluate(Xva, labels)})

base_df = pd.DataFrame(baseline_rows).set_index("baseline")

print("=== ผลลัพธ์ Baseline ===")
display(base_df.round(4))
```

```python
# 5. เปรียบเทียบ MiniBatchKMeans, KMeans และ GMM
K_RANGE = range(3, 13)
rows = []

print("กำลังเทรนและเปรียบเทียบโมเดล...")
for k in K_RANGE:
    # 5.1) MiniBatchKMeans (Primary Model)
    t = time.time()
    mbk_a = MiniBatchKMeans(n_clusters=k, batch_size=2048, n_init=3, random_state=42).fit(Xtr)
    mbk_b = MiniBatchKMeans(n_clusters=k, batch_size=2048, n_init=3, random_state=7).fit(Xtr)

    labels_a = mbk_a.predict(Xva)
    labels_b = mbk_b.predict(Xva)
    rows.append({
        "model": "MiniBatchKMeans", "k": k, **evaluate(Xva, labels_a),
        "stability_ARI": adjusted_rand_score(labels_a, labels_b), "sec": time.time() - t
    })

    # 5.2) KMeans (Standard Benchmark)
    t = time.time()
    km_a = KMeans(n_clusters=k, n_init=3, random_state=42).fit(Xtr)
    km_b = KMeans(n_clusters=k, n_init=3, random_state=7).fit(Xtr)

    labels_a = km_a.predict(Xva)
    labels_b = km_b.predict(Xva)
    rows.append({
        "model": "KMeans", "k": k, **evaluate(Xva, labels_a),
        "stability_ARI": adjusted_rand_score(labels_a, labels_b), "sec": time.time() - t
    })

    # 5.3) Gaussian Mixture Model (Probabilistic Alternative)
    t = time.time()
    gmm_a = GaussianMixture(n_components=k, covariance_type="diag", reg_covar=1e-3, n_init=1, random_state=42).fit(Xtr)
    gmm_b = GaussianMixture(n_components=k, covariance_type="diag", reg_covar=1e-3, n_init=1, random_state=7).fit(Xtr)

    labels_a = gmm_a.predict(Xva)
    labels_b = gmm_b.predict(Xva)
    rows.append({
        "model": "GMM", "k": k, **evaluate(Xva, labels_a),
        "stability_ARI": adjusted_rand_score(labels_a, labels_b), "sec": time.time() - t
    })

    print(f"k={k} เสร็จแล้ว")

res = pd.DataFrame(rows)
res.to_csv("tuning_results.csv", index=False)
print("รัน Tuning เสร็จสิ้น บันทึก tuning_results.csv แล้ว")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import linear_sum_assignment
from scipy.spatial.distance import cdist
from sklearn.cluster import KMeans
from sklearn.metrics import (silhouette_score, silhouette_samples,
                             davies_bouldin_score, calinski_harabasz_score,
                             adjusted_rand_score)
from IPython.display import display

_need = ["final_model", "BEST_K", "X_train_preprocessed", "X_val_preprocessed",
         "X_test_preprocessed", "train_idx", "val_idx", "test_idx", "fe", "meta"]
_missing = [v for v in _need if v not in globals()]
assert not _missing, f"ยังไม่ได้รันเซลล์ที่สร้างตัวแปร: {_missing}"

K = int(BEST_K)
Xtr = np.asarray(X_train_preprocessed, dtype=np.float32)
Xva = np.asarray(X_val_preprocessed,   dtype=np.float32)
Xte = np.asarray(X_test_preprocessed,  dtype=np.float32)

# --- ตรวจว่าใช้ index ชุดล่าสุด (70/15/15) และไม่ซ้อนกัน ---
assert len(Xtr) == len(train_idx) and len(Xva) == len(val_idx) and len(Xte) == len(test_idx), \
    "ขนาด X ไม่ตรงกับ index: ตรวจว่า train/val/test_idx มาจาก Section 8 รอบล่าสุด"
assert len(np.intersect1d(train_idx, test_idx)) == 0 and len(np.intersect1d(val_idx, test_idx)) == 0
assert final_model.n_clusters == K and final_model.n_features_in_ == Xte.shape[1]
assert np.isfinite(Xte).all(), "พบ NaN/inf ใน test"

mid_tr = meta["match_id"].to_numpy()[train_idx]
mid_te = meta["match_id"].to_numpy()[test_idx]
assert len(np.intersect1d(mid_tr, mid_te)) == 0, "พบแมตช์ซ้ำระหว่าง train กับ test"

# --- ใช้โมเดลที่ fit บน train เท่านั้น (ไม่ fit ใหม่) ---
labs = {"train": final_model.predict(Xtr),
        "val":   final_model.predict(Xva),
        "test":  final_model.predict(Xte)}
Xs = {"train": Xtr, "val": Xva, "test": Xte}

print(f"โมเดล: KMeans k={K} | Train {len(Xtr):,} | Val {len(Xva):,} | Test {len(Xte):,} แถว")
print(f"แมตช์ใน Test: {len(np.unique(mid_te)):,} | ซ้อนกับ Train: 0")
```

```python
def cluster_metrics(X, labels, n_other=200_000, n_sil=20_000, seed=42):
    """สุ่มชุดย่อยขนาดเท่ากันทุก split เพื่อให้เทียบกันได้ (silhouette O(n^2))"""
    rng = np.random.default_rng(seed)
    idx = rng.choice(len(X), min(n_other, len(X)), replace=False)
    Xq, lq = X[idx], labels[idx]
    return {
        "silhouette":        silhouette_score(Xq, lq, sample_size=min(n_sil, len(Xq)), random_state=seed),
        "davies_bouldin":    davies_bouldin_score(Xq, lq),
        "calinski_harabasz": calinski_harabasz_score(Xq, lq),
        "inertia_per_row":   -final_model.score(X) / len(X),     # ยิ่งต่ำยิ่งแนบ centroid
    }

rows = {s: cluster_metrics(Xs[s], labs[s]) for s in ["train", "val", "test"]}
tbl = pd.DataFrame(rows).T
tbl.loc["test − train"] = tbl.loc["test"] - tbl.loc["train"]
tbl.loc["test vs train (%)"] = (tbl.loc["test"] / tbl.loc["train"] - 1) * 100
print("เปรียบเทียบคุณภาพการจัดกลุ่ม (ชุดย่อย 200,000 แถว/ชุด, silhouette จาก 20,000 แถว)")
display(tbl.round(4))
test_metrics = rows["test"]
```

จากตารางเปรียบเทียบคุณภาพการจัดกลุ่ม สรุปได้ว่าโมเดลไม่มีปัญหา Overfitting และสามารถนำไปประยุกต์ใช้กับข้อมูลที่ไม่เคยเห็นมาก่อน (Generalization) ได้อย่างดีเยี่ยม

**คะแนนมีความเสถียรสูง:** ค่าตัวชี้วัดทั้งหมดได้แก่ Silhouette, Davies-Bouldin, Calinski-Harabasz และ Inertia ในชุดข้อมูล Train, Val และ Test มีค่าที่ใกล้เคียงกันมาก   

**ความคลาดเคลื่อนต่ำมาก:** เมื่อพิจารณาในแถว test vs train (%) พบว่าความแตกต่างของประสิทธิภาพระหว่างข้อมูลชุด Test และ Train มีค่าน้อยกว่า 1% ในทุกตัวชี้วัด (ต่างกันสูงสุดเพียง 0.4905% ใน Calinski-Harabasz)

```python
K_RANGE = np.arange(K)
prop = pd.DataFrame({s: np.bincount(labs[s], minlength=K) / len(labs[s]) * 100
                     for s in ["train", "val", "test"]}, index=K_RANGE)
prop["|test − train|"] = (prop["test"] - prop["train"]).abs()
tvd = 0.5 * (prop["test"] - prop["train"]).abs().sum()      # total variation distance (% จุด)
print("สัดส่วนสมาชิกของแต่ละ cluster (%)")
display(prop.round(2))
print(f"ผลต่างสูงสุด = {prop['|test − train|'].max():.2f} จุดเปอร์เซ็นต์ | "
      f"Total variation distance = {tvd:.2f}%")

fig, ax = plt.subplots(figsize=(10, 3.8))
w = 0.27
for j, s in enumerate(["train", "val", "test"]):
    ax.bar(K_RANGE + (j - 1) * w, prop[s], width=w, label=s)
ax.set_xticks(K_RANGE); ax.set_xlabel("Cluster"); ax.set_ylabel("% of events")
ax.set_title("Cluster size distribution: train vs val vs test"); ax.legend()
plt.tight_layout(); plt.show()
```

จากผลลัพธ์ในภาพ สามารถสรุปได้ว่า สัดส่วนการกระจายตัวของข้อมูลในแต่ละกลุ่ม (Cluster Distribution) มีความเสถียรและสม่ำเสมอสูงมาก ในทุกชุดข้อมูล

รายละเอียดหลักที่สนับสนุนข้อสรุปนี้ ได้แก่:

**ความสอดคล้องของสัดส่วน:** สัดส่วนของเหตุการณ์ในแต่ละ Cluster (0-9) ระหว่างชุด Train, Val และ Test มีค่าใกล้เคียงกันแทบจะสมบูรณ์แบบ ซึ่งเห็นได้ชัดเจนจากความสูงของแท่งกราฟทั้ง 3 สีที่อยู่ในระดับเดียวกันในทุกๆ กลุ่ม   

**ผลต่างสูงสุด (Maximum Difference) ต่ำมาก:** เมื่อเทียบความแตกต่างของสัดส่วนระหว่างชุด Test กับ Train พบว่าจุดที่คลาดเคลื่อนมากที่สุดมีผลต่างเพียง 0.23 จุดเปอร์เซ็นต์ (ที่ Cluster 5)   ค่าความคลาดเคลื่อนรวม (TVD) ต่ำเป็นพิเศษ: ค่า Total Variation Distance ของทั้งระบบอยู่ที่ 0.44% ซึ่งถือว่าต่ำมากๆ   

**สรุปในเชิงการวิเคราะห์:**
ผลลัพธ์นี้เป็นข้อพิสูจน์ที่แข็งแรงว่า โครงสร้างการเคลื่อนที่ (Movement Archetypes) ทั้ง 10 รูปแบบที่โมเดลค้นพบนั้น เป็นแพตเทิร์นธรรมชาติที่เกิดขึ้นจริงในเกมฟุตบอล ไม่ใช่แพตเทิร์นที่โมเดลจำเพาะเจาะจงมาแค่จากชุด Train (ไม่มี Overfitting ทางโครงสร้าง) โมเดลสามารถนำไปจัดกลุ่มเหตุการณ์ในแมตช์ใหม่ๆ ได้อย่างเสถียรและไม่ลำเอียง

```python
# (a) fit KMeans ใหม่บน test (ใช้เพื่อวัดความเสถียรของโครงสร้างเท่านั้น ไม่ใช่เพื่อเลือกโมเดล)
km_te = KMeans(n_clusters=K, n_init=3, random_state=7).fit(Xte)
ari_refit = adjusted_rand_score(labs["test"], km_te.labels_)          # ARI ไม่ขึ้นกับลำดับเลข cluster

# จับคู่ centroid train <-> test-refit ด้วย Hungarian แล้ววัดระยะ
C_tr, C_te = final_model.cluster_centers_, km_te.cluster_centers_
cost = cdist(C_tr, C_te)
r, c = linear_sum_assignment(cost)
matched = cost[r, c]
d_between = cdist(C_tr, C_tr); d_between[np.diag_indices(K)] = np.inf
print(f"(a) ARI ระหว่างโมเดล train กับโมเดล refit บน test: {ari_refit:.4f}")
print(f"    ระยะ centroid ที่จับคู่กัน: เฉลี่ย {matched.mean():.3f} | สูงสุด {matched.max():.3f} "
      f"(เทียบกับระยะ centroid ใกล้สุดระหว่างกันใน train เฉลี่ย {d_between.min(axis=1).mean():.3f})")

# (b) สุ่ม 80% ของแมตช์ใน test ซ้ำ 20 รอบ -> ความแปรปรวนระดับแมตช์
rng = np.random.default_rng(42)
uniq_m = np.unique(mid_te)
boot = []
for b in range(20):
    chosen = rng.choice(uniq_m, size=int(0.8 * len(uniq_m)), replace=False)
    m = np.isin(mid_te, chosen)
    boot.append(cluster_metrics(Xte[m], labs["test"][m], n_other=100_000, n_sil=10_000, seed=b))
boot = pd.DataFrame(boot)
summ = boot.agg(["mean", "std", "min", "max"]).T
print("\n(b) Metrics เมื่อสุ่ม 80% ของแมตช์ใน test ซ้ำ 20 รอบ")
display(summ.round(4))
```

1. ความเสถียรของโครงสร้าง (Structural Stability - ส่วน a)

- ค่า ARI เท่ากับ 0.5965: ค่า ARI (Adjusted Rand Index) นี้อยู่ในระดับที่พอใช้ได้ (ปานกลาง) แสดงให้เห็นว่าโครงสร้างการจัดกลุ่มที่ได้จากข้อมูล Train กับการเทรนใหม่บนข้อมูล Test มีความสอดคล้องกันพอสมควร แม้จะไม่สูงมากจนถึงระดับดีเยี่ยม (เช่น >0.8) แต่ก็ยืนยันได้ว่ารูปแบบที่หาเจอนั้นไม่ใช่เรื่องบังเอิญ

- ระยะห่างของ Centroid: ระยะห่างเฉลี่ยของ Centroid ที่จับคู่กันระหว่างสองโมเดลอยู่ที่ 0.454 (สูงสุด 0.784) ซึ่งน้อยกว่าระยะห่างระหว่างกลุ่มในชุด Train เอง (เฉลี่ย 2.180) อย่างเห็นได้ชัด ข้อมูลนี้มีความสำคัญมาก เพราะมันพิสูจน์ว่าจุดศูนย์กลางของสไตล์การเล่น (Archetypes) ที่โมเดลเรียนรู้จากชุด Train นั้น "คงที่" และนำไปประยุกต์ใช้กับข้อมูล Test ได้อย่างแม่นยำ

2. ความแปรปรวนระดับแมตช์ (Match-Level Variance - ส่วน b)

- เมื่อทดสอบสุ่มแมตช์ 80% จำนวน 20 รอบ พบว่าค่า Standard Deviation (std) ของทุก Metric ต่ำมากๆ (เช่น Silhouette std = 0.0017)

- ความแปรปรวนที่ต่ำมากระดับนี้เป็นการยืนยันความเสถียรของโมเดลเมื่อต้องเจอกับแมตช์ใหม่ๆ โมเดลสามารถทำงานได้อย่างคงเส้นคงวา ไม่ได้เก่งเฉพาะแมตช์ใดแมตช์หนึ่ง

สรุปภาพรวม:
โมเดล K-Means ($K=10$) มีความน่าเชื่อถือสูง สามารถค้นหารูปแบบ Movement Archetypes ที่มีความเสถียรและคงที่ (Robust) พร้อมนำไปใช้อธิบายหรือวิเคราะห์ผลในระดับแมตช์ได้อย่างมั่นใจครับ

```python
rng = np.random.default_rng(42)
is_pass_col = Xte.shape[1] - 1                      # is_pass เป็นคอลัมน์สุดท้ายของ ColumnTransformer
pass_lab = (Xte[:, is_pass_col] > 0.5).astype(int)   # is_pass ผ่านแค่ imputer ใน Section 9 จึงยังเป็น 0/1
rand_lab = rng.integers(0, K, size=len(Xte))
# random ที่คงสัดส่วนขนาด cluster เท่าโมเดลจริง (baseline ที่เข้มกว่า random เฉยๆ)
rand_sized = rng.permutation(labs["test"])

base = {
    f"KMeans k={K} (โมเดลที่เลือก)": labs["test"],
    "Pass/Carry (k=2)":             pass_lab,
    f"Random uniform (k={K})":      rand_lab,
    f"Random คงสัดส่วน cluster (k={K})": rand_sized,
}
base_tbl = pd.DataFrame({n: cluster_metrics(Xte, l) for n, l in base.items()}).T.drop(columns="inertia_per_row")
print("เทียบกับ baseline บน test (หมายเหตุ: Silhouette ของ k ต่างกันเทียบตรง ๆ ได้จำกัด ให้ดู random ที่ k เท่ากันเป็นหลัก)")
display(base_tbl.round(4))
```

**มีโครงสร้างชัดเจน ไม่ได้เกิดจากการเดาสุ่ม:** โมเดลหลักมีประสิทธิภาพทิ้งห่างการจัดกลุ่มแบบ Random ทั้งสองรูปแบบอย่างสิ้นเชิงในทุกตัวชี้วัด โดยมีค่า Silhouette Score เป็นบวก (0.1934) ในขณะที่กลุ่มเดาสุ่มมีค่าติดลบ (-0.0098 และ -0.0084) รวมถึงค่าการเกาะกลุ่ม (Calinski-Harabasz) ที่สูงถึงหลักหมื่นเทียบกับกลุ่มสุ่มที่ได้เพียงหลักหน่วย   

**ให้มิติการวิเคราะห์ที่ลึกกว่าการแยกแค่ Pass/Carry:** แม้การแบ่งข้อมูลเป็น 2 กลุ่มตามประเภทเหตุการณ์ ($K=2$) จะเป็นบรรทัดฐานพื้นฐาน แต่โมเดล $K=10$ กลับทำคะแนนได้ดีกว่าอย่างชัดเจน โดยมีค่าการแยกแยะกลุ่ม (Davies-Bouldin) ที่ต่ำกว่าเกือบครึ่ง (1.4195 เทียบกับ 2.6741) พิสูจน์ว่าการจัดกลุ่มด้วยพิกัดและทิศทางแบบ 10 Archetypes สามารถอธิบายความแตกต่างทางแท็กติกในสนามได้ดีกว่าการแยกด้วยประเภทแอ็กชันเพียงอย่างเดียว

```python
D = final_model.transform(Xte)                          # ระยะถึงทุก centroid
d1 = D.min(axis=1)
d2 = np.partition(D, 1, axis=1)[:, 1]
ratio = d1 / np.where(d2 == 0, np.nan, d2)           # ใกล้ 1 = อยู่กึ่งกลางระหว่าง 2 cluster
AMBIG = 0.80
ambig = ratio > AMBIG

# silhouette รายจุด (ชุดย่อย 10,000 แถว เพื่อไม่ให้ใช้ RAM มาก)
rng = np.random.default_rng(42)
sidx = rng.choice(len(Xte), min(10_000, len(Xte)), replace=False)
sil = np.full(len(Xte), np.nan)
sil[sidx] = silhouette_samples(Xte[sidx], labs["test"][sidx])

err = pd.DataFrame({"cluster": labs["test"], "d1": d1, "ambig": ambig, "sil": sil})
by_c = err.groupby("cluster").agg(
    size_pct=("cluster", lambda s: len(s) / len(err) * 100),
    dist_to_centroid=("d1", "mean"),
    ambiguous_pct=("ambig", lambda s: s.mean() * 100),
    mean_silhouette=("sil", "mean"),
    neg_silhouette_pct=("sil", lambda s: (s.dropna() < 0).mean() * 100),
)
print(f"Error analysis รายกลุ่ม | กำกวม = ระยะถึง centroid ที่ใกล้สุด / ใกล้อันดับสอง > {AMBIG}")
display(by_c.round(2))
print(f"ภาพรวม: กำกวม {ambig.mean()*100:.1f}% | silhouette < 0: {(sil[sidx] < 0).mean()*100:.1f}% "
      f"(ของแถวตัวอย่างสำหรับ silhouette)")

# --- แยกตามกลุ่มย่อยของข้อมูล ---
te_fe = fe.iloc[test_idx].reset_index(drop=True)
te_meta = meta.iloc[test_idx].reset_index(drop=True)
sub = pd.DataFrame({
    "type":   te_meta["type"].astype(str),
    "zone":   pd.cut(te_fe["start_x"], [-0.1, 40, 80, 120.1], labels=["Defensive", "Middle", "Final"]),
    "result": te_meta["result"].astype(str),
    "ambig":  ambig,
})
for col in ["type", "zone", "result"]:
    t = sub.groupby(col, observed=True)["ambig"].agg(["size", "mean"])
    t["mean"] *= 100
    t.columns = ["จำนวน events", "กำกวม (%)"]
    print(f"\nสัดส่วนจุดกำกวมแยกตาม {col}")
    display(t.round(2))

# กราฟ
fig, ax = plt.subplots(1, 2, figsize=(12, 3.6))
by_c["mean_silhouette"].plot.bar(ax=ax[0], color="tab:blue")
ax[0].axhline(0, color="k", lw=0.8); ax[0].set_title("Mean silhouette per cluster (test)")
by_c["ambiguous_pct"].plot.bar(ax=ax[1], color="tab:orange")
ax[1].set_title(f"% ambiguous points per cluster (ratio > {AMBIG})")
plt.tight_layout(); plt.show()


# %% [Cell 7] โปรไฟล์ cluster บน Test สอดคล้องกับ Train หรือไม่ (ความหมายของ archetype ไม่เปลี่ยน)
def profile(idx, lab):
    f = fe.iloc[idx].reset_index(drop=True)
    m = meta.iloc[idx].reset_index(drop=True)
    d = pd.DataFrame({"cluster": lab, "pass_pct": f["is_pass"].to_numpy() * 100,
                      "dist": m["dist"].to_numpy(), "start_x": f["start_x"].to_numpy(),
                      "dx": m["dx"].to_numpy(), "dy": m["dy"].to_numpy()})
    g = d.groupby("cluster")
    out = g[["pass_pct", "dist", "start_x"]].mean()
    out["direction_deg"] = np.degrees(np.arctan2(g["dy"].mean(), g["dx"].mean()))
    return out

p_tr, p_te = profile(train_idx, labs["train"]), profile(test_idx, labs["test"])
cmp_ = p_tr.join(p_te, lsuffix="_train", rsuffix="_test")
for c_ in ["pass_pct", "dist", "start_x", "direction_deg"]:
    cmp_[f"Δ{c_}"] = cmp_[f"{c_}_test"] - cmp_[f"{c_}_train"]
cmp_["Δdirection_deg"] = (cmp_["Δdirection_deg"] + 180) % 360 - 180      # ผลต่างมุมแบบวงกลม (-180..180)
print("โปรไฟล์ cluster: train vs test (Δ = test − train)")
display(cmp_.round(2))
print(f"ผลต่างสูงสุด -> pass% {cmp_['Δpass_pct'].abs().max():.2f} | dist {cmp_['Δdist'].abs().max():.2f} หลา | "
      f"start_x {cmp_['Δstart_x'].abs().max():.2f} หลา | ทิศ {cmp_['Δdirection_deg'].abs().max():.2f} องศา")
```

```python
# ==========================================================
# Cell 2 (ฉบับรวมตารางสรุป + กราฟสนามและลูกศร)
# ==========================================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from sklearn.cluster import KMeans
from IPython.display import display

N_ARROWS = 150
SHOW_START_HEATMAP = True

# ---------- 1) เตรียมข้อมูล ----------
Xtr_all = np.asarray(X_train_preprocessed)
if "final_model" not in globals():
    final_model = KMeans(BEST_K, n_init=3, random_state=42).fit(Xtr_all)
labels = final_model.predict(Xtr_all)

raw = df_clean.iloc[train_idx].reset_index(drop=True)
assert len(raw) == len(labels), "จำนวนแถวไม่ตรงกัน: ตรวจว่า train_idx เป็นของข้อ 8 ล่าสุด"

viz = pd.DataFrame({
    "cluster": labels,
    "start_x": raw["start_x"].to_numpy(), "start_y": raw["start_y"].to_numpy(),
    "end_x":   raw["end_x"].to_numpy(),   "end_y":   raw["end_y"].to_numpy(),
    "is_pass": (raw["type"] == "Pass").to_numpy(),
})
viz["dx"] = viz.end_x - viz.start_x
viz["dy"] = viz.end_y - viz.start_y
viz["dist"] = np.hypot(viz.dx, viz.dy)

# ---------- 2) ตารางสรุปพร้อมตั้งชื่ออัตโนมัติ ----------
g = viz.groupby("cluster")
summary = pd.DataFrame({
    "size_%":      g.size() / len(viz) * 100,
    "pass_%":      g["is_pass"].mean() * 100,
    "dist_yd":     g["dist"].mean(),
    "start_x":     g["start_x"].mean(),
    "start_y":     g["start_y"].mean(),
    "forward_yd":  g["dx"].mean(),
    "sideways_yd": g["dy"].mean(),
})
summary["direction_deg"] = np.degrees(np.arctan2(g["dy"].mean(), g["dx"].mean()))

def describe(r):
    kind = "Pass" if r["pass_%"] >= 50 else "Carry"
    length = "short" if r["dist_yd"] < 8 else ("medium" if r["dist_yd"] < 25 else "long")
    a = r["direction_deg"]; aa = abs(a); side = " (y+)" if a > 0 else " (y-)"
    if aa <= 22.5:     d = "forward"
    elif aa <= 67.5:  d = "diag-forward" + side
    elif aa <= 112.5: d = "sideways" + side
    elif aa <= 157.5: d = "diag-back" + side
    else:              d = "backward"
    zone = "own third" if r["start_x"] < 40 else ("middle" if r["start_x"] < 80 else "final third")
    return f"{kind}, {length}, {d}, from {zone}"

summary["label"] = summary.apply(describe, axis=1)
display(summary.round(1))
summary.to_csv("cluster_summary.csv")

# ---------- 3) ฟังก์ชันวาดสนามฟุตบอล ----------
def draw_pitch(ax, color="#9e9e9e"):
    ax.add_patch(Rectangle((0, 0), 120, 80, fill=False, ec=color, lw=1.5))
    ax.plot([60, 60], [0, 80], color=color, lw=1.5)
    for x0 in (0, 102):
        ax.add_patch(Rectangle((x0, 18), 18, 44, fill=False, ec=color, lw=1.5))
    ax.set_xlim(-2, 122); ax.set_ylim(-2, 82)
    ax.set_aspect("equal"); ax.axis("off")

clusters = sorted(viz["cluster"].unique())
ncols = 5
nrows = int(np.ceil(len(clusters) / ncols))

# ---------- 4) กราฟแสดงลูกศรพฤติกรรม (Quiver Plot) ----------
fig, axes = plt.subplots(nrows, ncols, figsize=(4.6 * ncols, 3.6 * nrows))
for c, ax in zip(clusters, axes.ravel()):
    d = viz[viz.cluster == c]
    s = d.sample(min(N_ARROWS, len(d)), random_state=1)
    draw_pitch(ax)
    ax.quiver(s.start_x, s.start_y, s.dx, s.dy, angles="xy", scale_units="xy",
              scale=1, width=0.004, alpha=0.35, color="tab:blue")
    ax.quiver(d.start_x.mean(), d.start_y.mean(), d.dx.mean(), d.dy.mean(),
              angles="xy", scale_units="xy", scale=1, width=0.014, color="red")
    ax.set_title(f"Cluster {c} | {summary.loc[c, 'size_%']:.1f}%\n{summary.loc[c, 'label']}", fontsize=9)
for ax in axes.ravel()[len(clusters):]:
    ax.axis("off")
plt.suptitle("Movement archetypes (attack left to right | blue = sample events, red = cluster mean)", fontsize=13)
plt.tight_layout()
plt.savefig("cluster_arrows.png", dpi=150)
plt.show()

# ---------- 5) Heatmap จุดเริ่มต้นเหตุการณ์ (Start Heatmap) ----------
if SHOW_START_HEATMAP:
    fig, axes = plt.subplots(nrows, ncols, figsize=(4.6 * ncols, 3.6 * nrows))
    for c, ax in zip(clusters, axes.ravel()):
        d = viz[viz.cluster == c]
        d = d.sample(min(100_000, len(d)), random_state=1)
        ax.hist2d(d.start_x, d.start_y, bins=[24, 16], range=[[0, 120], [0, 80]], cmap="YlOrRd", cmin=1)
        draw_pitch(ax)
        ax.set_title(f"Cluster {c}: where events start", fontsize=9)
    for ax in axes.ravel()[len(clusters):]:
        ax.axis("off")
    plt.tight_layout()
    plt.show()
```

```python
# 3) ตาราง ทีม-แมตช์ : สัดส่วน (%) ของแต่ละ archetype
ev = pd.DataFrame({
    "match_id": meta["match_id"].to_numpy(),
    "team": meta["team"].astype(str).to_numpy(),
    "result": meta["result"].astype(str).to_numpy(),
    "split": split_all,
    "cluster": lab_all,
})
cnt = (ev.groupby(["match_id", "team", "result", "split", "cluster"])
         .size().unstack("cluster", fill_value=0)
         .reindex(columns=range(K), fill_value=0))
cnt.columns = CL
tm = cnt.reset_index()
tm["n_events"] = tm[CL].sum(axis=1)
tm[CL] = tm[CL].div(tm["n_events"], axis=0) * 100
del ev, cnt

assert (tm.groupby("match_id")["team"].nunique() == 2).all(), "บางแมตช์ไม่มี 2 ทีม"
assert np.allclose(tm[CL].sum(axis=1), 100)
print(f"ทีม-แมตช์ทั้งหมด: {len(tm):,} | แมตช์: {tm['match_id'].nunique():,} | ทีม: {tm['team'].nunique():,}")
print(tm["result"].value_counts().to_string())

mix_by_result = tm.groupby("result")[CL].mean().reindex(["win", "draw", "loss"])
print("\n(A) สัดส่วน archetype เฉลี่ย (%) ต่อทีม-แมตช์ แยกตามผล")
display(mix_by_result.round(2))
```

```python
# ==========================================================
# Cell 4 (แบบแก้ไขขาด): เปรียบเทียบ Winner vs Loser
# ==========================================================
import warnings
import matplotlib as mpl

# ปิด Warning ทั้งหมดเกี่ยวกับฟอนต์
mpl.rcParams.update({'figure.max_open_warning': 0})
warnings.filterwarnings('ignore')

def bh(p):
    p = np.asarray(p, float); n = len(p); o = np.argsort(p)
    adj = p[o] * n / np.arange(1, n + 1)
    adj = np.minimum.accumulate(adj[::-1])[::-1]
    out = np.empty(n); out[o] = np.minimum(adj, 1)
    return out

def paired_table(tm_sub):
    dec = tm_sub[tm_sub["result"].isin(["win", "loss"])]
    w = dec[dec["result"] == "win"].set_index("match_id")
    l = dec[dec["result"] == "loss"].set_index("match_id")
    assert w.index.is_unique and l.index.is_unique
    common = w.index.intersection(l.index)
    diff = w.loc[common, CL] - l.loc[common, CL]
    n_diff = w.loc[common, "n_events"] - l.loc[common, "n_events"]
    rows = []
    for c in CL:
        d = diff[c].to_numpy(); n = len(d)
        sd = d.std(ddof=1); se = sd / np.sqrt(n)
        try:
            p = stats.wilcoxon(d).pvalue
        except ValueError:
            p = np.nan
        rows.append({"cluster": c, "mean_diff_pp": d.mean(),
                     "ci95_low": d.mean() - 1.96 * se, "ci95_high": d.mean() + 1.96 * se,
                     "effect_dz": d.mean() / sd if sd > 0 else np.nan, "p_wilcoxon": p})
    out = pd.DataFrame(rows).set_index("cluster")
    out["p_BH"] = bh(out["p_wilcoxon"].fillna(1).to_numpy())
    return out, len(common), n_diff

res_all, n_all, ndiff_all = paired_table(tm)
res_test, n_test, _ = paired_table(tm[tm["split"] == "test"])

print(f"\n(B) ผู้ชนะ − ผู้แพ้ ในแมตช์เดียวกัน | ทุกแมตช์: {n_all:,} | เฉพาะ test: {n_test:,}")
display(res_all.round(4))
print("เฉพาะ test:")
display(res_test.round(4))

# พล็อตกราฟ (ใช้ CL แทน TICK เพื่อตัดปัญหาเรื่องฟอนต์และตัวอักษรทับกัน)
fig, ax = plt.subplots(figsize=(10, 4))
x = np.arange(K); w_ = 0.38
for j, (res_, lab) in enumerate([(res_all, f"All Matches (n={n_all:,})"), (res_test, f"Test Split (n={n_test:,})")]):
    err = np.vstack([res_["mean_diff_pp"] - res_["ci95_low"], res_["ci95_high"] - res_["mean_diff_pp"]])
    ax.bar(x + (j - 0.5) * w_, res_["mean_diff_pp"], width=w_, yerr=err, capsize=3, label=lab)

ax.axhline(0, color="k", lw=0.8)
ax.set_xticks(x)
ax.set_xticklabels(CL, fontsize=11, fontweight='bold')
ax.set_xlabel("Archetype Cluster ID", fontsize=11)
ax.set_ylabel("Win − Loss (Percentage Points)", fontsize=11)
ax.set_title("Archetype Proportion Difference: Winner vs Loser (95% CI)", fontsize=12)
ax.legend(fontsize=10)
plt.tight_layout()
plt.show()
```

**สรุป: โมเดลที่เลือกเหมาะสมเพราะ**

1. KMeans (k=10) ดีกว่า baseline บน test ที่ไม่เคยเห็น ทั้งการแบ่งตาม Pass/Carry (Silhouette 0.1934 เทียบ 0.1098, Davies–Bouldin 1.4195 เทียบ 2.6741) และการสุ่ม (Silhouette ติดลบ)

2. ผลคงที่ข้ามชุดข้อมูลและข้าม seed (Silhouette 0.1932 / 0.1942 / 0.1934 ใน train / val / test, ARI ข้าม seed 0.9937, สัดส่วนกลุ่มต่างกัน TVD 0.44%) และไม่ใช้ผลแข่งขันหรือชื่อทีมสร้างกลุ่ม จึงไม่พบสัญญาณ data leakage

3. แต่ละกลุ่มมีโปรไฟล์เชิงแท็กติกที่ตีความได้และสอดคล้องกับ EDA (เช่น Cluster 5 คือลูกยาว 49 หลาจากแดนหลัง, Cluster 7 และ 8 คือพาบอลสั้น ≈3 หลา)

4. ข้อจำกัดที่ต้องยอมรับ:
   - Silhouette 0.19 ต่ำ และจุดกำกวม 36%
   - ARI เมื่อ refit บน test เพียง 0.60 ขอบเขตกลุ่มจึงเลื่อนได้เมื่อเปลี่ยนข้อมูล
   - Silhouette ของ KMeans แทบเท่ากันทุก k (k=3: 0.1831, k=10: 0.1923) การเลือก k=10 จึงมีหลักฐานเชิงตัวเลขอ่อน ที่เลือกเพราะตีความได้ละเอียดกว่า

*(2) ทีมที่ชนะมีสไตล์การเล่นอย่างไร*

เปรียบเทียบสัดส่วน archetype ของผู้ชนะกับผู้แพ้ในแมตช์เดียวกัน (3,082 แมตช์ที่มีผู้ชนะ ตัดแมตช์เสมอ; เฉพาะ test 449 แมตช์) ได้ความต่างที่ตรงกันทั้งชุดรวมและ test (p ที่ปรับ BH < 0.05 และเครื่องหมายเดียวกัน) ใน 6 กลุ่ม:

- **ผู้ชนะใช้มากกว่า:** Cluster 6 (+0.79 pp), 4 (+0.39), 8 (+0.33), 1 (+0.30) ซึ่งเป็นการจ่ายถอย/ข้างในแดนกลางและการพาบอลสั้น
- **ผู้ชนะใช้น้อยกว่า:** Cluster 5 (−1.44 pp) และ Cluster 3 (−0.82) ซึ่งเป็นลูกยาวตรงและจ่ายทแยงขึ้นหน้าจากแดนหลัง



**สรุป:** ทีมที่ชนะมักเล่นแบบครองบอลและหมุนบอลในแดนกลางมากกว่า และพึ่งลูกยาวจากแดนหลังน้อยกว่า อย่างไรก็ตามขนาดผลระดับแมตช์เล็ก (|dz| สูงสุด 0.29) สัดส่วน archetype จึงบอกแนวโน้ม ไม่ใช่ตัวแยกผู้ชนะจากผู้แพ้ได้แม่นยำ

**ผลลัพธ์นำไปใช้ประโยชน์ได้อย่างไร และโดยใคร**

- **โค้ช/สเกาต์:** ใช้ 10 archetypes เป็นภาษากลางอธิบายสไตล์ทีมและเตรียมรับมือคู่แข่ง เช่น ทีมที่พึ่งลูกยาวจากแดนหลัง (Cluster 5) สูงอาจถูกกดดันที่ต้นทางได้
- **นักวิเคราะห์ข้อมูลสโมสร:** นำ pipeline (cleaning → feature → scaler → KMeans) ไปใช้กับข้อมูล event ลีกอื่น และเทียบสไตล์ทีมด้วยมาตรฐานเดียวกัน
- **นักกีฬา/ทีมงาน:** ใช้แผนภาพลูกศรและ heatmap จุดเริ่มต้นของแต่ละ archetype ทำความเข้าใจรูปแบบการเคลื่อนที่ในแต่ละโซน
- **ข้อควรระวัง:** ใช้เพื่ออธิบายและเปรียบเทียบสไตล์ ไม่ใช่ทำนายผลแข่งหรือสรุปว่า archetype ใด "ทำให้ชนะ"

**ข้อจำกัดของข้อมูลและโมเดล**

1. **โครงสร้างกลุ่มอ่อน:** Silhouette 0.19, จุดกำกวม 36.1% (Carry 41.9%), ARI เมื่อ refit บน test 0.60 ขอบเขตกลุ่มจึงไม่คงที่เด็ดขาด
2. **เลือก k ไม่ชัด:** Silhouette แทบเท่ากันทุก k และ k=9 กับ k=10 ต่างกันเพียง 0.0023 จำนวนกลุ่มจึงเป็นการตัดสินใจเชิงตีความส่วนหนึ่ง
3. **ใช้พิกัดปลายทาง:** feature ทราบค่าหลังเหตุการณ์จบ จึงอธิบายสิ่งที่เกิดแล้วเท่านั้น ใช้ทำนายก่อนเหตุการณ์ไม่ได้
4. **ความสัมพันธ์ ไม่ใช่สาเหตุ:** ทีมที่ชนะบ่อยมักแข็งแกร่งกว่าและครองบอลมากกว่า สัดส่วน archetype อาจสะท้อนระดับทีมหรือสถานการณ์เกม (เช่น ทีมที่ตามอยู่มักเล่นบอลยาว) ซึ่งไม่ได้ควบคุมในโมเดล
5. **ผลระดับทีมกับระดับแมตช์ไม่ตรงกันทุกกลุ่ม:** เช่น Cluster 9 ผู้ชนะใช้น้อยกว่าเล็กน้อยในระดับแมตช์ (−0.30 pp) แต่ทีมที่ชนะบ่อยใช้มากกว่า (+0.37 pp) จึงสรุปเฉพาะรูปแบบที่ตรงกันทั้งสองระดับ
6. **ข้อมูลไม่เป็นตัวแทนทั้งโลก:** StatsBomb Open Data เป็นรายการที่ถูกเลือกมาเปิดเผย (ทั้งทีมชายและหญิง) บางทีมมีข้อมูลเพียง 30 แมตช์ และทีมที่ชนะบ่อยที่สุดในข้อมูลไม่ใช่ทีมที่ชนะบ่อยที่สุดในโลก
7. **ตัดข้อมูลบางส่วน:** เก็บเฉพาะ Pass และ Carry (69.7% ของข้อมูลดิบ 9.3 ล้านแถว) และตัด Carry ระยะ 0 จำนวน 354,970 แถว การเคลื่อนที่ที่ไม่มีบอลไม่อยู่ในข้อมูล
8. **ตัดแมตช์เสมอ:** การเปรียบเทียบผู้ชนะ–ผู้แพ้ตัดแมตช์เสมอ (879 จาก 3,961 แมตช์) ออก
9. **สมมติฐานของ KMeans:** ชอบกลุ่มทรงกลมและขนาดใกล้กัน การใช้ sin/cos ร่วมกับ log ระยะลดปัญหานี้ได้บางส่วน
